## 1. `if` 语句

判断报销发票是否属于日常费用类型：

````python
# 用元组类型表示日常费用类型
dly_exps = ("移动话费", "水电费", "市内车费")
# 用列表类型表示收到的报销发票
sbmt_fp = ["水电费", "住宿费", "培训费"]
# 用 if 语句判断要报销的发票是否为日常费用
if sbmt_fp[0] in dly_exps:
    print(sbmt_fp[0], "属于日常费用")
````

## 2. `if-else` 语句

判断发票属于专项费用还是日常费用：

````python
# 用元组类型表示日常费用类型
dly_exps = ("移动话费", "水电费", "市内车费")
# 用元组类型表示专项费用类型
spe_fee = ("培训费", "招待费", "市场活动费")
# 用列表类型表示收到的报销发票
sbmt_fp = ["市场活动费", "水电费", "培训费", "招待费"]
# 用 if 语句判断要报销的发票是否为专项费用
if sbmt_fp[0] in spe_fee:
    print(sbmt_fp[0], "属于专项费用")
else:
    print(sbmt_fp[0], "属于日常费用")
````

## 3. `if-elif-else` 语句

多类型逐一判断：

````python
# 用元组类型表示日常费用类型
dly_exps = ("移动话费", "水电费", "市内车费")
# 用元组类型表示专项费用类型
spe_fee = ("培训费", "招待费", "市场活动费")
# 用元组类型表示差旅费用类型
trvl_fee = ("住宿费", "机票费", "高铁费")
# 用列表类型表示收到的报销发票
sbmt_fp = ["住宿费", "市场活动费", "水电费"]
# 用 if 语句判断要报销的发票属于哪类费用
if sbmt_fp[0] in dly_exps:
    print(sbmt_fp[0], "属于日常费用")
elif sbmt_fp[0] in spe_fee:
    print(sbmt_fp[0], "属于专项费用")
elif sbmt_fp[0] in trvl_fee:
    print(sbmt_fp[0], "属于差旅费用")
````

## 6. `while` 语句

用计数器循环处理每张报销单：

````python
# 用元组定义网络费用的细分类
Ntfe_typ = ("网络安全维护费", "数据费", "云租赁费", "短信费")
# 用列表来定义报销单
sbmt_fp = ["数据费", "移动话费", "短信费"]
# 调用 len 函数来计算提交了多少张报销单
len_sbmt = len(sbmt_fp)
# 设置循环的初始值从 0 开始
i = 0
# 当循环次数小于提交的报销单数量时，就执行
while i < len_sbmt:
    # 判断列表中的报销单类型是否在元组的元素中
    if sbmt_fp[i] in Ntfe_typ:
        # 在元组的元素中，就判断为网络费
        print(sbmt_fp[i], "的费用类型为：网络费")
    else:
        # 不在元组的元素中，就判断为其他费用
        print(sbmt_fp[i], "的费用类型为：其他费用")
    # 循环次数加 1，然后跳转回 while 语句
    i = i + 1
print("——报销单类型判断完毕——")
````

## 7. `while-else` 语句

预算额度报销：余额足够则扣减报销，余额不足时进入 `else` 分支提示：

````python
# 预算额度
budget_amount = 5
# 初始余额等于预算额度
cur_balance = budget_amount
# 定义相关月份及费用支出
month_lst = [1, 2, 4, 7]
monthly_expenses = [1.2, 1.4, 2.3, 1.6]
# 设置累计报销费用的初始值
expense_total = 0
i = 0
# 提取要报销的费用
cur_expense = monthly_expenses[i]
# 余额大于等于要报销的费用时，就循环
while cur_balance >= cur_expense:
    # 允许报销，扣减预算余额
    cur_balance = cur_balance - cur_expense
    print(month_lst[i], "月报销费用：", cur_expense, \
          "预算额度变更为：", "{:.2f}".format(cur_balance))
    # 累计报销额度
    expense_total = expense_total + cur_expense
    i = i + 1
    cur_expense = monthly_expenses[i]
else:
    # 预算余额不足，给出提示
    print(month_lst[i], "月预算余额不足，不允许报销！")
    # 提示共报销了多少费用，剩余多少额度
    print("累计报销费用：", "{:.2f}".format(expense_total), \
          "剩余额度为：", "{:.2f}".format(cur_balance))
````

## 9. `break` 和 `continue` 语句（while 版）

`continue` 跳过小额费用单，`break` 在累计超 5000 元时退出：

````python
# 部门的费用数据
expenses = [500, 800, 1000, 1200, 900, 700, 1100, 1300, 600, 1000]
# 总费用初始化为 0
total_expenses = 0
i = 0

# while 循环用于统计报销单
while i < len(expenses):
    # 如果费用小于 1000，跳回开始处，继续循环
    if expenses[i] < 1000:
        i += 1
        continue
    # 统计单笔超过 1000 元的费用单的累计金额
    total_expenses += expenses[i]
    # 如果累计金额超过了 5000，退出循环
    if total_expenses > 5000:
        print("单笔大于1000元的报销单，累计金额已超过5000元！")
        break
    i += 1
````

## 10. `for` 语句

直接遍历报销单列表，无需计数器：

````python
# 用元组定义网络费用的细分子类
Ntfe_typ = ("网络安全维护费", "数据费", "云租赁费", "短信费")
# 用列表来定义报销单
sbmt_fp = ["数据费", "移动话费", "短信费"]
# 循环遍历所有的报销单
for i in sbmt_fp:
    # 判断列表中的报销单类型是否在元组的元素中
    if i in Ntfe_typ:
        # 在元组的元素中，就判断为网络费
        print(i, "的费用类型为：网络费")
    else:
        # 不在元组的元素中，就判断为其他费用
        print(i, "的费用类型为：其他费用")
print("——报销单类型判断完毕——")
````

## 11. `for-else` 语句

循环正常结束后进入 `else` 分支，计算平均成本：

````python
# 产品生产成本列表
costs = [120, 124, 118, 131, 125]
# 初始化总成本
total_cost = 0
# 累加每个成本项
for cost in costs:
    total_cost += cost
# 累计完毕，就计算平均成本
else:
    average_cost = total_cost / len(costs)
    print("产品生产成本总计为：", total_cost)
    print("产品生产成本平均为：", average_cost)
````

## 12. `break` 和 `continue` 语句（for 版）

与第 9 节 while 版逻辑一致，`for` 版无需手动维护索引 `i`：

````python
# 部门的费用数据
expenses = [500, 800, 1000, 1200, 900, 700, 1100, 1300, 600, 1000]
# 总费用初始化为 0
total_expenses = 0

# for 循环用于统计报销单
for i in expenses:
    # 如果费用小于 1000，跳过本次循环
    if i < 1000:
        continue
    # 统计单笔超过 1000 元的费用单的累计金额
    total_expenses += i
    # 如果累计金额超过了 5000，退出循环
    if total_expenses > 5000:
        print("单笔大于1000元的报销单，累计金额已超过5000元！")
        break
````

## 14. 任务：销售奖金包计算

用字典定义各区销售额，`for` 遍历 keys，按销售额区间适用不同提成比例（≤50 无奖金；≤200 提 2%；≤500 提 5%；≤1000 提 8%；其余提 10%）：

````python
# 用字典定义每个区的销售额
SalesInfo = {"华东区":580, "华南区":110, "东北区":44, \
"西南区":180, "中部区":350, "华西区":440}
# 用 for 循环调用字典中的 key
for i in SalesInfo.keys():
    # 通过 key，调用对应的值
    sales = SalesInfo[i]
    # 根据值的大小，判断其适用的提成比例
    if sales <= 50:
        bonus = 0
        print(i + "奖金包", bonus, "万元")
    elif sales <= 200:
        bonus = sales * 0.02
        print(i + "奖金包", bonus, "万元")
    elif sales <= 500:
        bonus = sales * 0.05
        print(i + "奖金包", bonus, "万元")
    elif sales <= 1000:
        bonus = sales * 0.08
        print(i + "奖金包", bonus, "万元")
    else:
        bonus = sales * 0.1
        print(i + "奖金包", bonus, "万元")
````
