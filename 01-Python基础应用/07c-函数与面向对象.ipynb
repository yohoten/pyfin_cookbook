## 题目 1：判断是否超过预算（函数定义）

定义函数 `Judge_fee(fee, budget)`：费用不超过预算返回"没有超预算"，否则返回"超过预算"。

```python
def Judge_fee(fee, budget):
    """判断是否超过预算"""
    if fee <= budget:
        return "没有超预算"
    else:
        return "超过预算"
```

## 题目 3：按司龄计算奖金（必备参数）

`lenSrv` 为司龄、`bsSlyr` 为基本薪资：司龄 ≥10 年奖金 ×2，≥5 年 ×1.5，否则 ×1；打印员工信息并返回奖金。

```python
def Employee(name, age, lenSrv, bsSlyr):
    #lenSrv为司龄，bsSlyr为基本薪资
    #根据司龄来计算奖金bonus
    if lenSrv >= 10:
        bonus = bsSlyr * 2
    elif lenSrv >= 5:
        bonus = bsSlyr * 1.5
    else:
        bonus = bsSlyr
    print("|姓名", name, "|年龄", age, "|司龄", lenSrv, "\n"
          "|基本薪资", bsSlyr, "|奖金", bonus)
    return bonus
```

## 题目 4：默认参数（示例 3）

`deduction=0` 为默认参数：不传入时按 0 处理，传入时用实参值覆盖。

```python
#1-5 2.1 函数的参数和调用 示例3
#函数的默认参数
#定义函数：calculate_tax，有三个参数
#其中deduction为默认参数，默认为0
def calculate_tax(income, tax_rate, deduction=0):
    tax = income * tax_rate - deduction
    return tax

# 示例用法
income = 100  # 收入
tax_rate = 0.13  # 税率

# 无抵扣项时，直接调用函数
tax = calculate_tax(income, tax_rate)
print(f"无抵扣项，税款为: {tax:.2f}")

# 有抵扣项时，指定参数值
deduction = 5  # 抵扣项金额
tax = calculate_tax(income, tax_rate, deduction)
print(f"有抵扣项，税款为: {tax:.2f}，抵扣项: {deduction}")
```

## 题目 5：可变参数 `*args`（元组打包）

`*args` 把多余的**位置参数**打包成**元组**。按参数个数分派计算：2 个 → 毛利润；3 个 → 营业利润；4 个 → 再乘 `(1 - 税率)` 得净利润；其余抛 `ValueError`。

```python
def calculate_profit(*args):
    financial_data = []
    print("args=", args, end="")
    for arg in args:
        financial_data.append(float(arg))

    if len(financial_data) == 2:
        profit = financial_data[0] - financial_data[1]
    elif len(financial_data) == 3:
        profit = financial_data[0] - financial_data[1] - financial_data[2]
    elif len(financial_data) == 4:
        profit = (financial_data[0] - financial_data[1] - financial_data[2]) \
                 * (1 - financial_data[3])
    else:
        raise ValueError("参数列表长度必须为 2、3、4")

    return profit
```

原题仅给出函数定义，直接运行无输出。〔补充演示〕分别以 2、3、4 个参数调用（注意：函数内的 `print(..., end="")` 不换行，`args=` 与调用结果打印在同一行）：

```python
raw_profit = calculate_profit(1000, 500)
print("毛利润:", raw_profit)
Operating_profit = calculate_profit(1000, 500, 200)
print("营业利润:", Operating_profit)
net_profit = calculate_profit(1000, 500, 200, 0.25)
print("净利润:", net_profit)
```

## 题目 7：思考 1 —— 计算薪酬

基本薪资 6000 元，考勤得分 9.5、绩效得分 8.2（各满分 10 分）：
考勤奖金 = 基本薪资 × 0.2 × 得分 / 10，绩效奖金 = 基本薪资 × 0.3 × 得分 / 10。

```python
#1-5 2.1函数 思考1
def cal_salary(bas_salary, att_score, perf_score):  
    """  
    计算薪酬的函数  
    参数:  
    bas_salary: 基本薪资，单位为元  
    att_score: 考勤得分，满分10分  
    perf_score: 绩效得分，满分10分  
    返回:  
    salary: 薪酬，单位为元  
    """  
    attendance_bonus = bas_salary * 0.2 * att_score / 10  
    performance_bonus = bas_salary * 0.3 * perf_score / 10 
    salary = bas_salary + attendance_bonus + performance_bonus  
    return salary  
  
# 调用函数，计算薪酬  
bas_salary = 6000  # 基本薪资，单位为元  
att_score = 9.5  # 考勤得分  
perf_score = 8.2  # 绩效得分  
  
salary = cal_salary(bas_salary, att_score, perf_score)  
print(f"薪酬是: {salary}元")
```

## 题目 8：可变参数 `**kwargs`（字典打包）

`**args` 把多余的**关键字参数**打包成**字典**（Python 3.7+ 保持传入顺序）。仍按参数个数分派计算，与题目 5 的 `*args` 版本对照。

```python
def calculate_profit(**args):
    financial_data = []
    for key, value in args.items():
        financial_data.append(float(value))

    print("args=", args)

    if len(financial_data) == 2:
        profit = financial_data[0] - financial_data[1]
    elif len(financial_data) == 3:
        profit = financial_data[0] - financial_data[1] - financial_data[2]
    elif len(financial_data) == 4:
        profit = (financial_data[0] - financial_data[1] - financial_data[2]) \
                 * (1 - financial_data[3])
    else:
        raise ValueError("参数列表长度必须为 2、3、4")

    return profit

income = 1000
cost = 500
expenses = 200
tax_rate = 0.25
raw_profit = calculate_profit(income=income, cost=cost)
print("毛利润:", raw_profit)
Operating_profit = calculate_profit(income=income, cost=cost, expenses=expenses)
print("营业利润:", Operating_profit)
net_profit = calculate_profit(income=income, cost=cost, expenses=expenses, tax_rate=tax_rate)
print("净利润:", net_profit)
```

## 题目 10：类的继承

`class Employee(People)` 表示 Employee 继承 People：子类自动获得父类的属性和方法（`display_info` 未重写，直接继承使用）；子类 `__init__` 中通过 `People.__init__(self, ...)` 完成父类部分的初始化。

```python
class People:
    name = ""
    age = 0
    gender = ""

    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

    def display_info(self):
        print(f"姓名: {self.name}, 年龄: {self.age}, 性别: {self.gender}")

class Employee(People):
    department = "部门"
    post = "岗位"
    Year_of_Employ = 0

    def __init__(self, name, age, gender, department, post, Year_of_Employ):
        People.__init__(self, name, age, gender)
        self.department = department
        self.post = post
        self.Year_of_Employ = Year_of_Employ
        self.Seniority = 2024 - Year_of_Employ

    def display_Employee(self):
        print(f"姓名: {self.name}, 岗位: {self.post}, 司龄: {self.Seniority}")

Emp1 = Employee("张振江", 30, "男", "财务部", "会计", 2016)
Emp1.display_info()
Emp1.display_Employee()
```

## 题目 11：方法重写与 `super()`

子类定义与父类**同名**的方法即"重写"，实例调用时执行子类版本；`super(子类名, 实例).方法()` 可显式调用父类版本。

```python
class People:
    name = ""
    age = 0
    gender = ""

    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

    def display_info(self):
        print(f"姓名: {self.name}, 年龄: {self.age}, 性别: {self.gender}")

class Employee(People):
    department = "部门"
    post = "岗位"
    Year_of_Employ = 0

    def __init__(self, name, age, gender, department, post, Year_of_Employ):
        People.__init__(self, name, age, gender)
        self.department = department
        self.post = post
        self.Year_of_Employ = Year_of_Employ
        self.Seniority = 2024 - Year_of_Employ

    def display_Employee(self):
        print(f"姓名: {self.name}, 岗位: {self.post}, 司龄: {self.Seniority}")

    def display_info(self):   # 重写父类方法
        print(f"姓名: {self.name}, 部门: {self.department}, 岗位: {self.post}")

Emp1 = Employee("张振江", 30, "男", "财务部", "会计", 2016)
Emp1.display_info()                # 调用子类重写的方法
super(Employee, Emp1).display_info()  # 调用父类的方法
```

## 题目 12：多层继承

`Major` → `Employee` → `People` 三层继承链：子类 `__init__` 调用直接父类的 `__init__`，逐层向上传递。原题仅给出类定义，直接运行无输出。〔补充演示〕实例化 `Major` 并调用：

```python
class People:
    name = ""
    age = 0
    gender = ""

    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

    def display_info(self):
        print(f"姓名：{self.name}, 年龄：{self.age}, 性别：{self.gender}")

class Employee(People):
    department = "部门"
    post = "岗位"
    Year_of_Employ = 0

    def __init__(self, name, age, gender, department, post, Year_of_Employ):
        People.__init__(self, name, age, gender)
        self.department = department
        self.post = post
        self.Year_of_Employ = Year_of_Employ
        self.Seniority = 2024 - Year_of_Employ

    def display_Employee(self):
        print(f"姓名：{self.name}, 岗位：{self.post}, 司龄：{self.Seniority}")

    def display_info(self):
        print(f"姓名：{self.name}, 部门：{self.department}, 岗位：{self.post}")

class Major(Employee):
    major = "专业"

    def __init__(self, name, age, gender, department, post, Year_of_Employ, major):
        Employee.__init__(self, name, age, gender, department, post, Year_of_Employ)
        self.major = major

    def display_Employee(self):
        print(f"姓名：{self.name}, 专业：{self.major}")
```

```python
Mj1 = Major("张振江", 30, "男", "财务部", "会计", 2016, "会计学")
Mj1.display_Employee()
Mj1.display_info()   # 继承自 Employee 的重写版本
```

## 题目 13：私有属性与私有方法

属性/方法名以**双下划线**开头（`__cal_salary`、`_bas_salary`）为私有成员，只能在类内部访问。原题仅给出类定义，直接运行无输出。

〔补充演示〕实例化并调用私有方法。由于名称改写（name mangling），类外须用 `_Employee__cal_salary` 形式访问——这也是"私有"只是一种约定而非强制隔离的体现：

```python
class Employee:
    name = ""
    post = ""
    att_score = 0.0
    perf_score = 0.0
    Upper = 0
    Lower = 0
    _bas_salary = 0.0   # 基础薪资，私有属性
    _salary = 0.0       # 薪资，私有属性

    def __init__(self, name, post, att_score, perf_score, Upper, Lower):
        self.name = name
        self.post = post
        self.att_score = att_score
        self.perf_score = perf_score
        self.Upper = Upper
        self.Lower = Lower
        self._bas_salary = 0.0
        self._salary = 0.0

    def __cal_salary(self):   # 私有方法
        if self.post == "财务部":
            self._bas_salary = 8000
        elif self.post == "采购部":
            self._bas_salary = 7000
        elif self.post == "行政部":
            self._bas_salary = 5000
        else:
            self._bas_salary = 6000

        attendance_bonus = self._bas_salary * 0.2 * self.att_score / 10
        performance_bonus = self._bas_salary * 0.3 * self.perf_score / 10
        self._salary = self._bas_salary + attendance_bonus + performance_bonus
        return self._salary
```

```python
Emp = Employee("张振江", "财务部", 9.5, 8.2, 0, 0)
salary = Emp._Employee__cal_salary()   # 名称改写后的访问方式
print("薪资 =", salary)
```

## 题目 14：任务实施 —— 资产负债率判断

用类封装"资产负债率 = 负债 / 资产"，比率在 (0.4, 0.6) 区间判定为正常。对 2021–2023 三年数据循环实例化并判断。

```python
class Asset_liability:
    Asset = 0.0
    Liability = 0.0

    def __init__(self, Asset, Liability):
        self.Asset = Asset
        self.Liability = Liability

    def cal_Asset_liability_ratio(self):
        ratio = self.Liability / self.Asset
        if 0.4 < ratio < 0.6:
            return "资产负债率正常"
        else:
            return "资产负债率异常"

lst_year = [2021, 2022, 2023]
lst_asset = [14824743567, 15074434759, 19146994680]
lst_liability = [8668952922, 13424278936, 10641931399]

i = 0
while i < len(lst_year):
    Ast_Liab = Asset_liability(lst_asset[i], lst_liability[i])
    IsNormal = Ast_Liab.cal_Asset_liability_ratio()
    print(lst_year[i], IsNormal)
    i += 1
```

**结果解读：** 2021 年 ≈ 0.585、2023 年 ≈ 0.556，落在 (0.4, 0.6) 区间 → 正常；2022 年 ≈ 0.890，明显偏高 → 异常。
