# 7. 运算符

## 算术运算符与字符串运算

```python
# 字符串与整数相乘的示例
x = "a"
y = 3
print("x*y=", x*y)
# 取模 % 的示例
x = 34
y = 5
print("x%y=", x%y)
# 取整除的示例
x = 12
y = 5
print("正数 x//y=", x//y)
x = -12
y = 5
print("负数 x//y=", x//y)
```

**问题：为什么 `x*y` 的结果是 3 个 a，而不是 3a？**

答：在 Python 中，同一个符号（如 `*`）根据操作数的数据类型不同，会有不同的含义：
- 当 `*` 用于两个数字时（如 `3 * 4`）：执行数学乘法运算；
- 当 `*` 用于"字符串"和"整数"时（如 `"a" * 3`）：执行**序列重复（Sequence Repetition）**操作，解释器执行 `x * y` 的逻辑等同于 `"a" + "a" + "a"`。

**问题：为什么 `x%y` 的结果是 4？**

答：`%` 运算符执行的是**取余数**操作（34 除以 5 余 4），而不是乘法操作。

`//` 为整除运算符，结果取整数：正整数取最小值（向下取整），负整数同样向更小方向取（-12//5 = -3）。

## 比较运算符

## 赋值运算符

```python
# 赋值运算符应用的示例
S_标准价格 = 200
F_价格系数 = 1.2
S_标准价格 *= F_价格系数
print(S_标准价格)
```

**说明：** `*=`（乘等于运算符）用于将右边的值乘以左边的变量，并将结果赋值给左边的变量。上例中用 `F_价格系数` 1.2 乘以 `S_标准价格` 200，结果即 1.2 × 200 = 240.0。

## 逻辑运算符（短路求值）

```python
# a 不为 False，返回 b
a = 1
b = 2
print(bool(a), bool(b), (a and b))

# a 为 False，返回 a
a = 0
b = 9
print(bool(a), bool(b), (a and b))
```

**解释：**

- `bool(a)`：Python 中非 0 数字、非空字符串或列表等为 `True`，0 / 空值 / `None` 为 `False`。
- `a and b` 的**短路求值**规则：
  - 第一个值为 `True` → 不会只返回 `True`，而是直接返回第二个值（如 `1 and 2` 返回 2）；
  - 第一个值为 `False` → 直接返回第一个值，无需判断第二个值（如 `0 and 9` 返回 0）。

**总结：**

1. Python 中 `and` 运算符不是简单返回 True/False，而是遵循短路求值；
2. 数字的布尔判断规则：0 是 False，非 0 数字（正数 / 负数）都是 True；
3. 这段代码的核心是展示 `and` 运算符的「短路特性」，而非单纯的逻辑真假判断。
