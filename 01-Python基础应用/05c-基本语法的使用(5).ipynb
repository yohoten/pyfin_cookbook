## 1. 元组的索引与切片

元组与列表的索引、切片规则一致：正向从 0 开始，反向从 -1 开始，切片左闭右开。

````python
# tuple 位置索引示例
tuple1 = ("cost", "refunds", "Insure", "subinsu", "mgmg", "finance")
print(tuple1[1], tuple1[-3])
print(tuple1[1:5])
print(tuple1[-6:-3])
print(tuple1[-3:-6])
print(tuple1[:2])
print(tuple1[-2:])
print(tuple1[:])
````

````python
tuple2 = ("营业成本", "退保金", "", "分保费用", "销售费用", "管理费用", "财务费用")
print(tuple2[-4:4])
````

`tuple2[-4:4]`：`-4` 对应正向索引 3（"分保费用"），到索引 4（不含），故只取到 `('分保费用',)`。

## 2. 元组的"修改"与删除

元组本身**不可变**，"修改"实际是创建新对象（可通过 `id()` 观察）：

````python
# 修改 tuple1 整个对象
tuple1 = ("cost", "refunds", "Insure")
print("修改前id:", id(tuple1))
tuple2 = ("营业成本", "退保金")
tuple1 = tuple1 + tuple2
print(tuple1)
print("修改后id:", id(tuple1))
````

前后 `id` 不同，说明 `tuple1 = tuple1 + tuple2` 生成的是**新元组对象**，并非原地修改。

## 3. 元组的 `+` 与 `*` 运算

````python
# 元组的 + 运算示例
tuple1 = ("cost", "refunds", "Insure")
tuple2 = ("营业成本", "退保金")
tuple1 = tuple1 + tuple2
print(tuple1)
````

````python
# 元组的 * 运算示例
tuple1 = ('营业收入', '利息收入')
tuple2 = (200,)*3
tuple3 = tuple1 + tuple2
print(tuple3)
````

注意区分 `(200)*3` 与 `(200,)*3`——前者是整数乘法，后者才是元组重复：

````python
tuple2 = (200)*3
print(tuple2)
tuple2 = (200,)*3
print(tuple2)
````

## 4. 字典的创建

````python
# 创建字典的示例
dict1 = {}
print(dict1, type(dict1))
dict2 = dict(Income=30, Asset=200, cost=20)
print(dict2)
dict3 = dict()
print(dict3)
````

字典的 key 可以是字符串、数字等**不可变类型**：

````python
# 字典的 key 值示例
dict1 = {"营业收入":2000, "销售费用":300, 34:89.2}
print(dict1)
````

> 注：讲义下一例标注为"字典的 key 值示例"，代码实际是把字典放入列表（`dict1=[{...}]`），输出为一个**包含单个字典的列表**，原样保留：

````python
# 字典的 key 值示例
dict1 = [{"营业收入":2000, "销售费用":300, 34:89.2}]
print(dict1)
````

**key 不可重复**：重复赋值时，后面的值覆盖前面的值：

````python
# 字典的 key 值示例
dict1 = {"营业收入":2000, "销售费用":300, "营业收入":89.2}
print(dict1)
````

元组是不可变类型，可以作为 key：

````python
dict1 = {('收入','成本'): 3230, "费用": 100}
print(dict1)
````

## 5. 字典的访问

通过 `字典[key]` 访问对应的值：

````python
dict1 = {"营业收入":2000, "销售费用":300, "研发费用":89.2}
print(dict1["销售费用"])
````

````python
dict1 = {1:"营业收入", 2:"销售费用", 3:"研发费用"}
print(dict1[3])
````

## 6. 字典的增加、修改与删除

对**不存在的 key** 赋值即为新增，对**已存在的 key** 赋值即为修改：

````python
dict1 = {"营业收入":2000, "销售费用":300}
dict1["研发费用"] = 89.2
print(dict1)
````

````python
dict1 = {"营业收入":2000, "销售费用":300}
dict1["销售费用"] = 500
print(dict1)
````

`pop(key)` 删除指定键值对并返回其值；`clear()` 清空字典：

````python
dict1 = {"营业收入":2000, "销售费用":300}
dict1.pop("销售费用")
print(dict1)
````

````python
dict1 = {"营业收入":2000, "销售费用":300}
dict1.clear()
print(dict1)
````

### 会报错的示例：`del dict1` 后再访问（原样保留）


实际执行报错：`NameError: name 'dict1' is not defined`（`del dict1` 删除的是整个字典对象）。

````python
dict1 = {"Income":100, "cost":30}
dict1["asset"] = 200
dict1['cost'] = 50
dict1.pop("Income")
print(dict1)
````

## 7. 集合的创建

- `set(字符串)` 会把字符串**拆成单个字符**形成集合；
- `set(字典)` 取字典的**键**形成集合；
- `{元素}` 是集合字面量，但**空的 `{}` 是字典**不是集合。

````python
set1 = set("营业收入")
print(set1)
set2 = set({"收入":20, "资产":300})
print(set2)
set3 = {"Income"}
print(set3, type(set3))
````

````python
set2 = set(("收入", "资产"))
print(set2)
````

````python
set2 = {}; set3 = {"revenue"}
print(type(set2), type(set3))
````

输出为 `<class 'dict'> <class 'set'>`——空 `{}` 是字典。

## 8. 集合的两个特性：无序、去重

````python
set1 = {'Income', 'Asset', 'Cost', 'Revenue'}
print(set1)
````

````python
set1 = {'Income', 'Asset', 'Income', 'Cost', 'Revenue', 'Cost'}
print(set1)
````

重复元素自动去重。集合**无序**，因此不能用位置索引访问（输出顺序也可能与书写顺序不同，每次运行可能变化）。

### 会报错的示例：对集合做下标访问（原样保留）

```python
set1 = {'Income', 'Asset', 'Income', 'Cost', 'Revenue', 'Cost'}
print(set1[2])
```

实际执行报错：`TypeError: 'set' object is not subscriptable`。

## 9. 集合元素的增加与删除

````python
# 集合新增元素示例-add
set1 = set(("营业收入", "利息收入", "已赚保费"))
set1.add("利息收入")
print(set1)
set1.add("其他业务收入")
print(set1)
````

`add()` 对已存在元素不产生任何效果（去重）。`update()` 可同时合并多个可迭代对象：

````python
# 新增集合元素示例-update
set1 = set(("营业收入", "利息收入", "已赚保费"))
set1.update({1, 2})
print(set1)
set2 = {"Income", "Interest", "Revenue"}
set2.update([3, 4], (5, 6))
print(set2)
````

`remove(x)` 移除存在的元素：

````python
# 集合 remove() 方法 - 移除存在的元素
set2 = {"Income", "Interest", "Revenue"}
set2.remove("Income")
print(set2)
````

### 会报错的示例：remove 不存在的元素（原样保留）

```python
# 集合 remove() 方法 - 移除不存在的元素
set2 = {"Income", "Interest", "Revenue"}
set2.remove("cost")
print(set2)
```

实际执行报错：`KeyError: 'cost'`。

`discard(x)` 与 `remove` 类似，但移除**不存在的元素不报错**：

````python
# 集合 discard() 方法 - 移除存在的元素
set2 = {"Income", "Interest", "Revenue"}
set2.discard("Income")
print(set2)
# 集合 discard() 方法 - 移除不存在的元素
set2.discard("cost")
print(set2)
````

`pop()` **随机**弹出一个元素（无序），不能传参数；`clear()` 清空集合：

````python
# 集合 pop() 方法
set2 = {"Income", "Interest", "Revenue"}
print(set2)
set2.pop()
print(set2)
````

### 会报错的示例：`pop()` 带参数（原样保留）

```python

set2.pop("Revenue")
```

实际执行报错：`TypeError: set.pop() takes no arguments (1 given)`。

`clear()` 示例：

````python
# 集合 clear() 方法
set2 = {"Income", "Interest", "Revenue"}
print(set2)
set2.clear()
print(set2)
````

## 10. 集合运算

`-` 差集、`|` 并集、`&` 交集、`^` 对称差（并集减交集）：

````python
# 集合运算示例
A = set("Income")
B = set("Revenue")
print("A=", A)
print("B=", B)
print("A-B为：", A - B)
print("A|B为：", A | B)
print("A&B为：", A & B)
print("A^B为：", A ^ B)
````

## 11. 任务实施：库存周转率计算

用元组构建计算项目（名称），用列表构建各年数值，通过索引组合成计算公式；存货周转率 = 营业成本 ÷ 平均库存。

````python
# 用元组构建计算项目
fItem = ("营业成本", "年初库存", "年末库存")
# 用列表构建项目的值
fValue1 = [68.96, 22.9, 29.03]
fValue2 = [63.10, 29.03, 25.32]
fValue3 = [71.53, 25.32, 36.82]
fValue4 = [102, 36.82, 50.15]

# 通过列表索引的值构建计算公式
fIvt_trn1 = fValue1[0] / ((fValue1[1] + fValue1[2]) / 2)
fIvt_trn2 = fValue2[0] / ((fValue2[1] + fValue2[2]) / 2)
fIvt_trn3 = fValue3[0] / ((fValue3[1] + fValue3[2]) / 2)
fIvt_trn4 = fValue4[0] / ((fValue4[1] + fValue4[2]) / 2)

# 通过元组索引的值显示公式说明
print("计算公式：")
print("    库存周转率={Item0}/((({Item1}+{Item2})/2)".\
    format(Item0=fItem[0], Item1=fItem[1], Item2=fItem[2]))
print("----------------------------")
````

````python
# 格式化输出计算结果
print("计算结果：")
print("----------------------------")
print("  第1年 | 第2年 | 第3年 | 第4年  ")
print("----------------------------")
print("  {:.2f}".format(fIvt_trn1), "| {:.2f}".format(fIvt_trn2), \
      " | {:.2f}".format(fIvt_trn3), \
      " | {:.2f}".format(fIvt_trn4))
print("----------------------------")

# 简单分析库存周转率的变化趋势
print("数据分析：")
if fIvt_trn2 > fIvt_trn1:
    print("  第2年存货周转率为上升趋势")
else:
    print("  第2年存货周转率为下降趋势")
if fIvt_trn3 > fIvt_trn2:
    print("  第3年存货周转率为上升趋势")
else:
    print("  第3年存货周转率为下降趋势")
if fIvt_trn4 > fIvt_trn3:
    print("  第4年存货周转率为上升趋势")
else:
    print("  第4年存货周转率为下降趋势")
````
