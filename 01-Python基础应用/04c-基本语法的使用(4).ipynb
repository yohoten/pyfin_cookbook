## 1. 用 `format()` 格式化数字

财务数据输出常用四种格式：保留小数位、千分位分隔、百分比、科学计数法。

````python
# 将数字保留2位小数输出
x = 3.1415926
print("{:.2f}".format(x))

# 将长数字用逗号分隔后输出
x = 329685069385
print("{:,}".format(x))

# 将数字保留2位小数，按百分比输出
x = 0.329685069385
print("{:.2%}".format(x))

# 将数字保留2位小数，按指数格式输出
x = 329685069385
print("{:.2e}".format(x))
````

## 2. `format()` 的命名参数与字典解包

使用 `**字典` 将字典的键值对直接传给 `format()`，格式串中以 `{键名}` 引用：

````python
# format() 设置参数的示例
Income = {"name": "营业收入", "revenue": 58934068.0}
print("项目: {name}  值: {revenue}".format(**Income))
````

`{:02}` 表示不足 2 位时左侧补 0，适合时间、日期的输出：

````python
print('{:02}:{:02}:{:02}'.format(22, 2, 56))
````

格式串中还可用 `[索引]` 访问序列元素：

````python
list = ['营业收入', '利润表']
print('{names[0]} 是 {names[1]} 中的一个项目'.format(names=list))
````

## 3. 字符串索引与切片

- **正向索引**从 0 开始，**反向索引**从 -1 开始；
- 切片语法 `[起始:结束:步长]`，取**左闭右开**区间。

````python
Str1 = "Income"
Str2 = "利润表中文"

# 从左向右访问某个位置的字符
print("单字符-正向索引：", Str1[4], Str2[2], sep=" | ")

# 从右向左访问某个位置的字符
print("单字符-反向索引：", Str1[-3], Str2[-2], sep=" | ")

# 访问某一段位置的字符
print("多字符-正向索引：", Str1[0:4], Str2[1:3], sep=" | ")
print("多字符-省略写法：", Str1[:4], Str2[1:], Str2[:], sep=" | ")
print("多字符-反向索引：", Str1[-6:-3], Str2[-3:-1], sep=" | ")
print("多字符-从右向左：", Str1[-3:-6], Str2[-1:-3], sep=" | ")
````

**要点：**

- `Str1[-3:-6]`、`Str2[-1:-3]` 因起始索引在结束索引右侧，取不到内容，返回**空字符串**；
- 切片不会因越界报错，超出范围的部分自动截断。

````python
Str1 = "Income"
Str2 = "利润表中文"
print(Str1[4:4], Str2[1:6], Str2[0:-3], sep="，")
````

## 4. 字符串运算符：`+`、`*`、`in` / `not in`、`r`

````python
# + 运算符示例
Str2 = "利润表中文"
Str2 = Str2[:3] + "格式"
print(Str2)

# r/R 运算符示例
print(r'\n', '\n')
print("输出测试")
````

**要点：**

- `+` 为字符串拼接；
- `r'\n'` 为**原始字符串**，反斜杠不再转义，原样输出 `\n`；而 `'\n'` 是换行符。

````python
Str2 = "Income"
Str2 = Str2[:3] + "."
Str3 = Str2 * 2
print("m" not in Str3)
````

`Str2` 为 `"Inc."`，`Str3 = Str2 * 2` 得到 `"Inc.Inc."`，其中不含字符 `m`，故 `"m" not in Str3` 为 `True`。

## 5. 转义字符

````python
# 输出反斜杠、单引号、双引号
print("\\", "\'", "\"")
# 退格符 \b（删除前一个字符）
print("Hello\b World!")
# 原始字符串 + 空字符 \000 + 普通字符串
print(r"\000输出起始", "\000", "结尾")
# 制表符 \t（相当于按 Tab 键）
print("Hello \t World!")
# 回车符 \r（光标回到行首，覆盖后续内容）
print("HelloKitty\rWorld!")
````

````python
print("this is \\n ")
print("this is \n ")
````

`"\\n"` 输出字面量 `\n`；`"\n"` 则是换行符，因此第二个 `print` 先换行再输出内容。

`%s` 占位字符串、`%d` 占位整数，右侧用元组按顺序传入参数：

## 7. f-string（Python 3.6+）

在字符串前加 `f`，用 `{}` 直接嵌入变量或表达式：

````python
Dict1 = {'item': '营业收入', 'value': 3000}
print(f'Dict1["item"]: {Dict1["value"]}')
````

f-string 中的 `{}` 支持任意表达式：

````python
x = 2000
print(f'x+1000={x+1000}')
````

`{}` 中表达式的引号规则：f-string 外层使用单引号时，内部字典取值可用双引号 `Dict1["value"]`；若外层已用双引号，则内部需改用单引号，否则引号冲突。

下面是讲义中出现的**错误示例**（原样保留，仅作对照，不在本 Notebook 中执行）：

```python
Dict1 = {'item': '营业收入', 'value': 3000}
print(f'{Dict1("item")}: {Dict1("value")}')
```

`Dict1("item")` 把字典当成函数调用，会抛出 `TypeError: 'dict' object is not callable`。正确写法应为 `Dict1["item"]`。

f-string 末尾加 `=` 可同时输出**表达式源码与结果**，便于调试：

````python
x = 'this is test'
print(f'{x+" begin"=}')
````

## 8. 列表的索引与切片

列表的索引、切片规则与字符串一致：

````python
# list 位置索引示例
list1 = ["Income", "Interest", "Insure", "cost", "refunds", "develop"]
print(list1[1])
print(list1[-3])
print(list1[1:5])
print(list1[-6:-3])
print(list1[-3:-6])
print(list1[:-1])
print(list1[2:])
print(list1[:])
````

````python
list2 = ["营业收入", "利息收入", "已赚保费", "营业成本", "退保金", "研发费用"]
print(list2[2:-2])
````

`list2[2:-2]` = 从索引 2 到 -2（不含），即 `['已赚保费', '营业成本']`。

## 9. 列表元素的增加、删除与修改

| 操作 | 方法 | 说明 |
|---|---|---|
| 增加 | `append(x)` | 在**末尾**追加元素 |
| 增加 | `insert(i, x)` | 在索引 i 处插入元素 |
| 删除 | `del list[i]` | 按索引删除 |
| 删除 | `pop(i)` | 按索引删除并返回被删元素（默认删除最后一个） |
| 修改 | `list[i] = x` | 直接赋值修改 |

````python
# append 方法示例
list2 = ["营业收入", "利息收入"]
list2.append("研发费用")
print(list2)

# insert 方法示例
list2 = ["营业收入", "利息收入"]
list2.insert(0, "研发费用")
print(list2)
````

````python
# del 方法示例
list2 = ['营业收入', '利息收入', '研发费用']
del list2[1]
print(list2)

# pop 方法示例
list2 = ['营业收入', '利息收入', '研发费用']
list2.pop(0)
print(list2)
````

````python
# 修改列表元素的方法
list2 = ['营业收入', '利息收入', '研发费用']
list2[2] = '已赚保费'
print(list2)
````

````python
list1 = ['asset', 'revenue', 'cost']
list1.append("income")
print(list1)
list1.pop(1)
print(list1)
list1[0] = "liability"
print(list1)
````

## 10. 列表的 `+` 与 `*` 运算

`+` 为列表拼接，`*` 为列表重复：

````python
# 列表的 + 运算示例
list1 = ['营业收入', '利息收入']
list2 = [200, 30]
list3 = list1 + list2
print(list3)
````

````python
# 列表的 * 运算示例
list1 = ['营业收入', '利息收入']
list2 = [200] * 3
list3 = list1 + list2
print(list3)
````

`insert()` 的第二个参数可以是**列表本身**，此时形成嵌套列表：

````python
list1 = ['营业收入', '利息收入']
list2 = [10, 20] * 2
list1.insert(1, list2)
print(list1)
````

## 11. 嵌套列表的访问

`list3[0]` 取出子列表，`list3[0][1]` 再取其内部元素：

````python
list1 = ["营业收入", "利息收入", "已赚保费"]
list2 = [20, 30]
list3 = [list1, list2]
print(list3)
print(list3[0])
print(list3[0][1])
````

````python
list1 = ["revenue", "income", "cost"]
list2 = [10, 20]
list3 = [list2, list1]
print(list3[1][1])
````

## 12. `operator` 模块：序列比较

`operator.eq(a, b)` 等价于 `a == b`，逐元素比较列表内容：

````python
# 导入 operator 模块
import operator

list1 = [10, "income"]
list2 = [20, "revenue"]
list3 = [20, "revenue"]

print("list1与list2比较：", operator.eq(list1, list2))
print("list2与list3比较：", operator.eq(list2, list3))
````

````python
# 导入 operator 模块
import operator
list1 = ["book", 20]
list1[0] = 20
list2 = [20] * 2
print(operator.eq(list1, list2))
````

`list1` 首元素被改为 20 后为 `[20, 20]`，`list2` 为 `[20, 20]`，两者内容相等，故返回 `True`。
